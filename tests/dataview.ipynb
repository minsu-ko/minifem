```python
T = df.temperature_K
k = df.thermal_conductivity_W_per_cmK

plt.loglog(T, k, color='k')

# 주 눈금(major)과 보조 눈금(minor) 모두에 그리드 표시
plt.grid(True, which="both", ls="-", alpha=0.5)
plt.xlabel('Temperature (K)')
plt.ylabel('Thermal Conductivity (W/cm K)')
plt.title('constantan')
```
